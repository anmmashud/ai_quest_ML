### .read_csv()
- [Pandas `read_csv()` Documentation](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html)
|**Parameter**|**Explanation**|**Example**|
|-|-|-|
|`"file_path"`|Path or URL of the CSV file|"`Dhaka_Rent.csv`"|
|`header=`|Row number to use as the column names|`0`, `1`, `2`, `None`|
|`sep=""`|Column separator character|`,`,`;`,` `,`   `,|
|`index_col=`|Specific Column to use as row label|`0`, `1`, `2`, `False`|
|`usecols=[""]`|Read only selected columns|`["area"]`, `["rent"]`, `["area","rent"]`|
|`na_values=[""]`|dditional strings to recognize as `NA`/`NaN`|`['']`,`['?', ' ', '-']`|
|`encoding=""`|Prevents UnicodeDecodeError|`utf-8`|
|`na_filter=`|Detect missing value markers (empty strings and the value of na_values)|In data without any `NA` values, passing `na_filter=False` can improve the performance of reading a large file.|

### Code
There are two example. First one is simple version. and 2nd one is for understanding the code what working in background. it is with all necessery parameters with their default values.
both code will show the same result.

```python
# example one
import pandas as pd
df = pd.read_csv("Dhaka_Rent.csv", sep=";")
df.head()
```

```python
# Example 2
import pandas as pd
df = pd.read_csv("Dhaka_Rent.csv", header=0, index_col=False, sep=";", usecols=["area","rent"], encoding='utf-8')
df.head()
```



### .head()
- [Pandas `.head()` Documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html)

|**Parameter**|**Explanation**|**Example**|
|-|-|-|
|`DataFrame.head()`|Returns the first n rows (default n=5))|`df.head()`, `df.head(10)`,`df.head(-2)`|
|`DataFrame.tail()`|Returns the last n rows|`df..tail()`,`.tail(10)`|
|`DataFrame[start:stop]`|Rows from `start` index up to but not including `stop`|`df[3:7]`, `df[3:],`,`df[:7]`|
