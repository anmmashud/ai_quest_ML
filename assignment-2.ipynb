## Problem 1: Variable Swap
Write a Python program to swap the values of two variables
without using a temporary variable.

```python
# Take input of two values
a = input("Enter 1st Value: ")
b = input("Enter 2nd Value: ")

# Print the values before swaping
print(f"\nSwaping:\nBefore : a = {a}, b = {b}")

# Swaping the values
a, b = b, a

# Print the values after swaping
print(f"After  : a = {a}, b = {b}")
```

## Problem 5: Temperature Converter
Write a Python program that converts a temperature in
Celsius to Fahrenheit. Take the Celsius temperature as input from the user.

```python
temp_in = input("Enter temperature in Celsius (°C): ")

# this code will ask if any string given
while True:
    
    # if `try` gets True loop will break
    try:
        temp_c = float(temp_in)
        temp_f = (temp_c*(9/5)) + 32
        print(f"\n{temp_in}°C = {temp_f}°F")
        break
        
    # if the logic get False, it will run `except`
    except:
        temp_in = input("Input must be a number: ")
```

## Problem 7: String Palindrome
Write a Python function to check if a given string is a
palindrome or not.

```python
'''Sample Input
Strings: madam, level, WOW, did
Numbers: 121, 12321
'''
# Taking user input
text = input("Enter a word: ")

# an empty string
rev = ""

# loop for get each characters form input 
for i in text:
    rev = i + rev

# cheking if it's a palindrome or not
if text == rev:
    print(f'\n✅ "{text}" is a palindrome ✅')
else:
    print(f'\n❌ "{text}" is not a palindrome ❌')
```

## Problem 14: Grades Classification
Write a Python program that takes a student’s percentage as input and prints their corresponding grade according to the following criteria: 
- 90% or above: A+ 
- 80-89%: A 
- 70-79%: B 
- 60-69%: C 
- Below 60%: Fail

```python
# taking input from user (keep outside the loop)
user_input = input('Enter your marks: ')

# conditon to handle error if user gives any non integer number
while True:
    try:
        number = float(user_input)
        if number >= 90:
            print("Your got - A+")
        elif number >= 80:
            print("Your got - A")
        elif number >= 70:
            print("Your got - B")
        elif number >= 60:
            print("Your got - C")
        else:
            print("Your got - Fail")
        break
        
    # if user gives any string it will hit error and ask again
    except:
        user_input = input('Input must be an Integer: ')
```

## Problem 18: Quadratic Equation Solver
Write a Python program that takes the coefficients
(a, b, c) of a quadratic equation as input and calculates and prints the real roots (if
they exist) or a message indicating the complex roots.

1. **Linear Equation (1st degree)**
   * Form: ( ax + b = 0 )
   * Example: ( 2x + 3 = 0 )
   * Has **one solution**: ( x = -b/a )
2. **Quadratic Equation (2nd degree)**
   * Form: ( ax^2 + bx + c = 0 )
   * Example: ( x^2 - 5x + 6 = 0 )
   * Has **two solutions**, which may be real or complex.

3. **Cubic Equation (3rd degree)**
   * Form: ( ax^3 + bx^2 + cx + d = 0 )
   * Example: ( x^3 - 6x^2 + 11x - 6 = 0 )
   * Can have **one or three real roots**.

4. **Quartic Equation (4th degree)**
   * Form: ( ax^4 + bx^3 + cx^2 + dx + e = 0 )
   * Example: ( x^4 - 10x^2 + 9 = 0 )
   * Can have **up to four real roots**.


```python
''' (a)x^2 + (b)x + (c) = 0 '''  
import math

a = int(input("Enter coefficient of a: "))
b = int(input("Enter coefficient of b: "))
c = int(input("Enter coefficient of c: "))

D = b**2 - 4*a*c

if D>0:
    x1 = (-b + math.sqrt(D)) / (2*a)
    x2 = (-b - math.sqrt(D)) / (2*a)
    print(f"1st real root is: {x1:.2f}\n2nd real root is: {x2:.2f}")
elif D==0:
    x0 = (-b) / (2*a)
    print(f"One read root is {x0}")
else:
    real_part = -b / (2*a)
    imaginary_part = math.sqrt(-D)/(2*a)
    # print(f"Complex roots: {real_part:.2f} + {imaginary_part:.2f}i and {real_part:.2f} - {imaginary_part:.2f}i")
    print(f"Complex roots: ({real_part} + {imaginary_part}i) and ({real_part} - {imaginary_part}i)")
```
