## Problem 1: Variable Swap
Write a Python program to swap the values of two variables
without using a temporary variable.

```python
# Take input of two values
a = input("Enter 1st Value: ")
b = input("Enter 2nd Value: ")

# Print the values before swaping
print(f"\nSwaping:\nBefore : a = {a}, b = {b}")

# Swaping the values
a, b = b, a

# Print the values after swaping
print(f"After  : a = {a}, b = {b}")
```

## Problem 2: Even or Odd
Write a Python program that takes an integer as input and prints
whether it is even or odd.


```python
# Taking input as an integer
num = int(input("Enter a Number: "))

# Deviding the number by 2
if num%2 == 0:
    print(f"{num} is an Even number")
else:
    print(f"{num} is an Odd number")
```

## Problem 3: String Reverse
Write a Python function to reverse a given string and return the
reversed string.

```python
user_input = input("Enter anything: ")

rev_str = ""

for char in user_input:
    rev_str = char + rev_str
print(f"Reverse string: {rev_str}")
```

## Problem 4: Type Conversion
Given a list of integers, write a Python program to convert each
element of the list to a string.

```python
# given int list
int_list = [1,2,3]

# conversion logic (int to str)
str_list = [str(i) for i in int_list]     # list comprehension

# print the result
# print(type(int_list[1]), "\n", type(str_list[1]))

# using `type` method
if type(int_list[1]) is int:
    print(f"{int_list} = Integer List")
else:
    print(f"{int_list} = String List")

# using `isinstance` method

if isinstance(str_list[1], int):
    print(f"{str_list} = Integer List")
else:
    print(f"{str_list} = String List")
```

## Problem 5: Temperature Converter
Write a Python program that converts a temperature in
Celsius to Fahrenheit. Take the Celsius temperature as input from the user.

```python
temp_in = input("Enter temperature in Celsius (°C): ")

# this code will ask if any string given
while True:
    
    # if `try` gets True loop will break
    try:
        temp_c = float(temp_in)
        temp_f = (temp_c*(9/5)) + 32
        print(f"\n{temp_in}°C = {temp_f}°F")
        break
        
    # if the logic get False, it will run `except`
    except:
        temp_in = input("Input must be a number: ")
```

## Problem 6: Data Type Checker
Write a Python function that takes a variable as input and
returns the data type of the variable as a string (e.g., “int”, “float”, “str”, “list”, etc.)

```python
# variable with various Data Types
a = 54
b = 323.55
c = "My name is Mashud"
d = {43,43,5646,75,5}
e = {"id":"123", 'name': 'Mashud'} 
f = [43,54, "abc", 123]
g = (343, 'a', 'b',53)


# definging a fucntion to print the name of a datatype
def get_type(var):
    return type(var).__name__

# showing all results
# the printing part also can be perform in loop
print("Data Types: \n")
print(f"{a} =>",get_type(a))
print(f"{b} =>",get_type(b))
print(f"{c} =>",get_type(c))
print(f"{d} =>",get_type(d))
print(f"{e} =>",get_type(e))
print(f"{f} =>",get_type(f))
print(f"{g} =>",get_type(g))
```

## Problem 7: String Palindrome
Write a Python function to check if a given string is a
palindrome or not.

```python
'''
Some Examples of Palindrome:
Stringss: madam, level, WOW, did
Numbers: 121, 12321
'''
# Taking user input
text = input("Enter a word: ")

# an empty string
rev = ""

# loop for get each characters form input 
for i in text:
    rev = i + rev

# cheking if it's a palindrome or not
if text == rev:
    print(f"✅ {text} is a palindrome ✅")
else:
    print(f"❌ {text} is not a palindrome ❌")
```
